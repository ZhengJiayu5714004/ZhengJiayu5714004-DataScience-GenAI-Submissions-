### Visualising the Data

```markdown
### Description of `0` and `1` in the Dataset
In the Titanic dataset, the `Survived` column is a binary classification variable:
* **`0`**: Represents **Deceased** (the passenger did not survive the disaster).
* **`1`**: Represents **Survived** (the passenger survived the disaster).

When we plot the 'Survival Rate' in the subsequent charts, it represents the proportion of passengers who survived (where `Survived` equals `1`).
```

````python
# Plot the relationship between ticket class (Pclass) and survival (Survived)
sns.barplot(data=students, x='Pclass', y='Survived', errorbar=None)
plt.xlabel('Pclass')
plt.ylabel('Survival Rate')
plt.title('Survival Rate by Ticket Class')
plt.show()
````

Interstingly if you parents had masters-level education your math score is lower! Let's try another:

````python
# Plot the relationship between Age and Survival Rate
fig, ax = plt.subplots(figsize=(8, 5))
sns.kdeplot(data=students, x='Age', hue='Survived', multiple='fill', common_norm=False, palette='Set1', alpha=0.8)
plt.xlabel('Age')
plt.ylabel('Survival Rate')
plt.title('Survival Rate by Age')
plt.show()
````

```markdown
As we can see, the probability of survival varies significantly across different age groups. For instance, younger children (on the far left) show a relatively higher survival rate.

We can also use a histogram to inspect the distribution shape of specific numeric features such as 'Fare':
```

````python
# Plot the distribution of ticket fares (Fare)
ax = sns.displot(students['Fare'].dropna(), kde=True)
plt.title('Distribution of Fare')
plt.show()
````

Whilst this is not exactly normal - slightly skewed - its probably fine in practice.

We can use the _pairplot( )_ method to visualise pairs of variables:

````python
# Plot pairwise relationships between Age, Fare, and Pclass
sns.pairplot(data=students[['Age', 'Fare', 'Pclass']].dropna(), hue='Pclass', diag_kind='hist')
plt.show()
````

```markdown
### Interpretation of the Pairplot (Age, Fare, and Pclass)
This pairplot visualizes the relationship between the passenger's Age, Fare, and Ticket Class (Pclass):

* **Pclass vs. Fare (Socio-Economic Division)**:
  There is a strong correlation between ticket class and fare. First-class passengers (`Pclass = 1`) paid significantly higher fares (extending up to £500), while third-class passengers (`Pclass = 3`) are heavily clustered at the lowest fare range (under £30).

* **Pclass vs. Age (Demographics)**:
  First-class passengers generally skew older, with a concentration of passengers in their 30s, 40s, and 50s. Third-class passengers are younger on average, primarily consisting of young adults in their 20s.

* **Age vs. Fare**:
  The highest fares were mostly paid by passengers in their 30s to 50s, which is expected as this age group typically had the most wealth to purchase luxury first-class tickets.
```

Here we have the scatter plot when comparing two variables, and when its only a single variable we have a histogram.
